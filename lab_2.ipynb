**7. Simple Rule-Based Agent**

```python
def weather_agent(weather):
    if weather == "sunny":
        return "Go outside";
    elif weather == "rainy":
        return "Take umbrella";
    elif weather == "snowy":
        return "Stay inside";
    else:
        return "Unknown weather";

print(weather_agent("sunny"));
```

```python
def battery_agent(battery):
    if battery < 20:
        return "Charge";
    else:
        return "Continue";

print(battery_agent(15));
```

```python
def traffic_agent(color):
    if color == "red":
        return "Stop";
    elif color == "yellow":
        return "Wait";
    elif color == "green":
        return "Go";
    else:
        return "Invalid color";

print(traffic_agent("green"));
```

```python
def eligibility_agent(marks, attendance):
    if marks >= 50 and attendance >= 75:
        return "Eligible";
    else:
        return "Not Eligible";

print(eligibility_agent(65, 80));
```

```python
def road_agent(road_open, traffic_heavy):
    if road_open == True and traffic_heavy == False:
        return "Main Road";
    else:
        return "Alternate Road";

print(road_agent(True, False));
```

```python
def room_agent(status):
    if status == "dirty":
        return "Suck";
    elif status == "clean":
        return "Move";
    else:
        return "Unknown status";

print(room_agent("dirty"));
```

**8. Vacuum Cleaner Agent**

```python
def vacuum_agent(location, status):
    if status == "dirty":
        return "Suck"
    elif status == "clean":
        return "Move"
    else:
        return "Unknown Status"

print(vacuum_agent("A", "dirty"));
print(vacuum_agent("A", "clean"));
print(vacuum_agent("B", "dirty"));
print(vacuum_agent("B", "clean"));
```

```python
def vacuum_agent(location, status):
    if status == "dirty":
        return "Suck"
    else:
        return "Move"

percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty"), ("B", "clean")];

for location, status in percepts:
    print(vacuum_agent(location, status));
```

```python
def vacuum_agent(location, status):
    if location != "A" and location != "B":
        return "Unknown Location"
    elif status == "dirty":
        return "Suck"
    else:
        return "Move"

print(vacuum_agent("C", "dirty"));
```

```python
def vacuum_agent(location, status):
    if status == "dirty":
        return "Suck"
    else:
        return "Move"

percepts = [("A", "dirty"), ("A", "clean"), ("B", "dirty"), ("B", "clean")];
actions = [];

for location, status in percepts:
    actions.append(vacuum_agent(location, status));

print(actions);
```
